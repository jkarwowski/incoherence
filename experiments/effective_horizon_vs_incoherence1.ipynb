```python
# Remove only extreme outliers using IQR method with larger threshold
def remove_outliers_iqr(data_x, data_y, data_z):
    # Convert to numpy arrays for easier manipulation
    arr_x = np.array(data_x)
    arr_y = np.array(data_y)
    arr_z = np.array(data_z)
    
    # Calculate IQR for y (using 3.0 for extreme outliers only)
    q1_y, q3_y = np.percentile(arr_y, [25, 75])
    iqr_y = q3_y - q1_y
    lower_y = q1_y - 1.5 * iqr_y
    upper_y = q3_y + 1.5 * iqr_y
    
    # Calculate IQR for z (using 3.0 for extreme outliers only)
    q1_z, q3_z = np.percentile(arr_z, [25, 75])
    iqr_z = q3_z - q1_z
    lower_z = q1_z - 1.5 * iqr_z
    upper_z = q3_z + 1.5 * iqr_z
    
    # Keep only points where both y and z are within bounds
    mask = (arr_y >= lower_y) & (arr_y <= upper_y) & (arr_z >= lower_z) & (arr_z <= upper_z)
    
    return arr_x[mask], arr_y[mask], arr_z[mask]

x_clean, y_clean, z_clean = remove_outliers_iqr(x, y, z)
print(f"Original data points: {len(x)}")
print(f"After removing outliers: {len(x_clean)}")
```
